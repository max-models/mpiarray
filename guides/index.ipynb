# Guides

Executable notebooks explain mpiarray from array creation through complete simulations. Read the rendered pages with their generated figures, or download a notebook to run and change it yourself. Longer setup and plotting cells can be expanded with **Show code**.

- [Distributed arrays](/mpiarray/guides/distributed-arrays/) — creation, storage, gathering and files.
- [Layouts](/mpiarray/guides/layouts/) — decomposition, ownership and redistribution.
- [Halo exchange](/mpiarray/guides/halo-exchange/) — boundaries, stencils and deposition.
- [Operations and reductions](/mpiarray/guides/operations/) — arithmetic, indexing and collective operations.
- [GPU arrays](/mpiarray/guides/gpu/) — CuPy and CUDA-aware MPI, with CPU-compatible examples.
- [PETSc](/mpiarray/guides/petsc/) — DMDAs, vector transfers and numbering.
- [Sparse stencil operators](/mpiarray/guides/sparse/) — fixed-pattern operators and PETSc matrices.
- [xarray](/mpiarray/guides/xarray/) — labels, coordinates and generated plots.
- [Using physical units](/mpiarray/guides/units/) — compatible arithmetic, assignments and reductions.
- [Heat equation](/mpiarray/guides/heat-equation/) — a complete stencil simulation.
- [Particles](/mpiarray/guides/particles/) — migration and deposition.
- [Building, reducing and saving arrays](/mpiarray/guides/working-with-data/) — rank-local data and I/O.

For notebook authoring and code-folding tags, see [Building the docs](/mpiarray/development/docs/).
